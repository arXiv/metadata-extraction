```python
default_output_path = "../data/2311_scopus_17416"
```

```python

import json
import os
import pandas as pd

# Load the CSV file
df = pd.read_csv(os.path.join(default_output_path, 'vip_institutions.csv'))

# Drop any missing values in the rorId column and collect unique entries as a set
ror_set = set(df['rorId'].dropna().unique())

# Step 2: Load the ground truth JSON file
with open(os.path.join(default_output_path, 'result_combined.json'), 'r') as f:
    ground_truth = json.load(f)

# Step 3: Filter each institution's ROR URLs to include only those in ror_set
filtered_ground_truth = {}
for institution_id, ror_list in ground_truth.items():
    filtered_urls = [url for url in ror_list if url in ror_set]
    if filtered_urls:  # Only include keys with at least one valid URL
        filtered_ground_truth[institution_id] = filtered_urls

# Optionally: Save the filtered mapping to a new JSON file
with open(os.path.join(default_output_path, 'vip_result_combined.json'), 'w') as f:
    json.dump(filtered_ground_truth, f, indent=4)

# Print the filtered result
# print(json.dumps(filtered_ground_truth, indent=4))
```
